```python
# removing html tags(not asked , doing it just for my practice) -> by using regex(regular expressions)
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'',text)
```

```python
def remove_stopwords(text):
    new_text =[]

    for i in text.split():
        if i in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(i)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)
```

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
import numpy as np
import gc

# Load stopwords ONCE outside the function ✅
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    new_text = []
    for i in text.split():
        if i in stop_words:        # uses pre-loaded set, not reloading every time ✅
            new_text.append('')
        else:
            new_text.append(i)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)

# Free RAM before processing
gc.collect()

# Apply in batches ✅
batch_size = 5000
chunks = np.array_split(df, len(df) // batch_size)

cleaned = []
for i, chunk in enumerate(chunks):
    cleaned.extend(chunk['review'].apply(remove_stopwords))
    print(f"Batch {i+1}/{len(chunks)} done ✅")

df['cleaned_review'] = cleaned
print("Done! ✅")
```
